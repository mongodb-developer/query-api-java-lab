# Basic CRUD: Insert

## What can the `insert` operation do?

The [`insert`](https://www.mongodb.com/docs/drivers/java/sync/current/crud/insert/?utm_campaign=devrel&utm_source=third-part-content&utm_medium=cta&utm_content=crud-operations-java-workshop&utm_term=ricardo.mello) operations let us add new documents to a MongoDB collection.

In this section, we will use operations such as `insertOne()` and `insertMany()` to create and store documents in the `books` collection.

## Startup code

This cell imports the MongoDB Java Driver, connects to MongoDB, and initializes the `library` database and `books` collection used in the insert examples.

`````java
// Import the MongoDB Driver using Maven
%maven org.mongodb:mongodb-driver-sync:5.5.1
%maven org.slf4j:slf4j-simple:1.7.36
    
import com.mongodb.client.MongoClient;
import com.mongodb.client.MongoClients;
import com.mongodb.client.MongoDatabase;
import com.mongodb.client.MongoCollection;
import com.mongodb.client.FindIterable;

import org.bson.Document;
import org.bson.json.JsonWriterSettings;
import org.bson.types.ObjectId;

import java.util.List;

// Configure SLF4J Simple Logger to suppress MongoDB driver logs in the notebook output.
System.setProperty("org.slf4j.simpleLogger.defaultLogLevel", "off");
System.setProperty("org.slf4j.simpleLogger.log.org.mongodb.driver", "off");

// Set your connection String
String connectionString = "mongodb://admin:mongodb@localhost:27017/";

// Define our database and collection. We'll use the `library` variable that points to our Database and `books` that points to the collection we're using.
MongoClient mongoClient = null;
try {
    // connect to MongoDB
    mongoClient = MongoClients.create(connectionString); 
} catch (Exception e) {
    System.out.println(e);
}

MongoDatabase library = mongoClient.getDatabase("library");
MongoCollection<Document> books = library.getCollection("books");
`````

## CRUD: Insert

### Insert one book

In this example, we create a new `Document` representing a book and use `insertOne()` to add it to the `books` collection.

`````java
Document elQuijote = new Document();
elQuijote.append("title", "El Quijote").append("year", 1500);

books.insertOne(elQuijote);
`````

### Find the inserted document

In this example, we query the `books` collection to verify that the document inserted with the year `1500` was successfully stored in MongoDB.

`````java
Document bookFrom1500 = new Document("year", 1500);

books.find(bookFrom1500)
     .forEach(b -> System.out.println("Book: " + b.toJson()));
`````

### Find the same document using the `_id`

Can you find the same document again, but this time using the `_id` field instead of the `year`?

Fix the code below by adding the correct `ObjectId`.

`````java
Document bookFrom1500 = new Document("_id", new ObjectId("")); // add the _id here!

books.find(bookFrom1500)
     .forEach(b -> System.out.println("Book: " + b.toJson()));
`````
