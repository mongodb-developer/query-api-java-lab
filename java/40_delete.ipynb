# Basic CRUD: Delete

## What can the `delete` operation do?

The [`delete`](https://www.mongodb.com/docs/drivers/java/sync/current/crud/delete/?utm_campaign=devrel&utm_source=third-part-content&utm_medium=cta&utm_content=crud-operations-java-workshop&utm_term=ricardo.mello) operations let us remove documents from a MongoDB collection.

In this section, we will use operations such as `deleteOne()` and `deleteMany()` to remove documents from the `books` collection.

## Startup code

```java
// Import the MongoDB Driver using Maven
%maven org.mongodb:mongodb-driver-sync:5.5.1
%maven org.slf4j:slf4j-simple:1.7.36
    
import com.mongodb.client.MongoClient;
import com.mongodb.client.MongoClients;
import com.mongodb.client.MongoDatabase;
import com.mongodb.client.MongoCollection;
import com.mongodb.client.FindIterable;
import com.mongodb.client.result.DeleteResult;

import static com.mongodb.client.model.Filters.eq;

import org.bson.conversions.Bson;
import org.bson.Document;

import java.util.List;

// Configure SLF4J Simple Logger to suppress MongoDB driver logs in the notebook output.
System.setProperty("org.slf4j.simpleLogger.defaultLogLevel", "off");
System.setProperty("org.slf4j.simpleLogger.log.org.mongodb.driver", "off");

// Set your connection String
String connectionString = "mongodb://admin:mongodb@localhost:27017/";

// Define our database and collection. We'll use the `library` variable that points to our Database and `books` that points to the collection we're using.
MongoClient mongoClient = null;
try {
    // connect to MongoDB
    mongoClient = MongoClients.create(connectionString); 
} catch (Exception e) {
    System.out.println(e);
}

MongoDatabase library = mongoClient.getDatabase("library");
MongoCollection<Document> books = library.getCollection("books");
```

## Insert one book

We want to make sure this book exists in the collection so we can later remove it using a delete operation.

```java
Document platero = new Document()
        .append("_id", "platero")
        .append("title", "Platero y yo")
        .append("year", 1900);

books.insertOne(platero);
```

### Find the inserted book

In this example, we query the `books` collection using the `_id` field to confirm that the `Platero y yo` document exists before deleting it.

```java
Document bookFrom1500 = new Document("_id", "platero"); 

books.find(bookFrom1500)
    .forEach(b -> System.out.println("Book: " + b.toJson())); 
```

### Delete one book

In this example, we use `deleteOne()` to remove the document with `_id` equal to `"platero"` from the `books` collection. 

The returned `DeleteResult` lets us verify how many documents were deleted.

```java
Document plateroFilter = new Document("_id", "platero");

DeleteResult deleteResult = books.deleteOne(plateroFilter);

// Check the number of deleted documents
long deletedCount = deleteResult.getDeletedCount();

System.out.println("Deleted " + deletedCount + " document(s).");
```
