# CRUD: Find and Project

## What can the `find()` operation do?

The [Find documents](https://www.mongodb.com/docs/drivers/java/sync/current/crud/query-documents/find/?utm_campaign=devrel&utm_source=third-part-content&utm_medium=cta&utm_content=crud-operations-java-workshop&utm_term=ricardo.mello) operation lets us query and retrieve documents from a MongoDB collection.

In this section, we will use `find()` with filters such as `eq()`, `gt()`, `gte()`, and `and()` to search for documents and apply conditions.

We will also use **projections** such as `fields()`, `include()`, and `excludeId()` to return only the fields we need.

## Find one book

This is the **Read** part of CRUD. We use [`find()`](https://www.mongodb.com/docs/drivers/java/sync/current/crud/query-documents/find/?utm_campaign=devrel&utm_source=third-part-content&utm_medium=cta&utm_content=crud-operations-java-workshop&utm_term=ricardo.mello) to query documents from a collection.

In this example, `find()` returns the query result, and `first()` retrieves the first matching document from the `books` collection.

```java
Document firstBook = books.find().first();

if (firstBook != null) {
    System.out.println("Book: " + firstBook.toJson());
} else {
    System.out.println("No book was found.");
}
```

# Find and Project

In this example, we use [`eq()`](https://www.mongodb.com/docs/manual/reference/operator/query/eq/?utm_campaign=devrel&utm_source=third-part-content&utm_medium=cta&utm_content=crud-operations-java-workshop&utm_term=ricardo.mello) to find the book with the title `"Treasure of the Sun"` and [`project`](https://www.mongodb.com/docs/manual/reference/operator/aggregation/project/?utm_campaign=devrel&utm_source=third-part-content&utm_medium=cta&utm_content=crud-operations-java-workshop&utm_term=ricardo.mello) to return only the `title`, `_id` and `pages` fields.

```java
var filter = eq("title", "Treasure of the Sun");
var projection = include("title", "pages");

books.find(filter)
    .projection(projection)
    .forEach(b -> System.out.println(b.toJson()));
```

## Find books written after 2021

In this example, we use [`gt()`](https://www.mongodb.com/docs/manual/reference/operator/query/gt/?utm_campaign=devrel&utm_source=third-part-content&utm_medium=cta&utm_content=crud-operations-java-workshop&utm_term=ricardo.mello), which means **greater than**, to filter books whose `year` is greater than `2021`, and project only the `title`, `_id` and `year`.

```java
filter = gt("year", 2021);
projection = include("title", "year");

books.find(filter)
    .projection(projection)
    .forEach(b -> System.out.println(b.toJson()));
```

## Find all books with 2000 pages or more

We use [`gte()`](https://www.mongodb.com/docs/manual/reference/operator/query/gte/?utm_campaign=devrel&utm_source=third-part-content&utm_medium=cta&utm_content=crud-operations-java-workshop&utm_term=ricardo.mello), which means **greater than or equal to**, to filter books whose `pages` value is `2000` or higher.

Then, we use a projection to return the `title`, `genres`, and `_id` fields.

```java
filter = gte("pages", 2000);
projection = include("title", "year", "genres");

books.find(filter)
    .projection(projection)
    .forEach(book -> System.out.println("Book: " + book.toJson()));
```

## Exclude the `_id` Field

In the previous example, MongoDB still returned the `_id` field because it is included by default, even when we do not explicitly list it in the projection.

To remove it, we use [`excludeId()`](https://www.mongodb.com/docs/drivers/java/sync/current/builders/projections/?utm_campaign=devrel&utm_source=third-part-content&utm_medium=cta&utm_content=crud-operations-java-workshop&utm_term=ricardo.mello). To combine `excludeId()` with `include()`, we wrap them with `fields()`.


```java
projection = fields(
    include("title", "pages"),
    excludeId()
);

books.find(filter)
    .projection(projection)
    .forEach(book -> System.out.println("Book: " + book.toJson()));
```
